Questo programma aprirà una finestra di disegno e disegnerà un quadrato nella finestra. A volte la finestra non sarà visibile subito. Potresti dover spostare la finestra del browser web per vedere l'area di disegno.

```python
import turtle
dan = turtle.Turtle()
dan.shape('turtle')

for i in range(4):
    dan.forward(50)
    dan.right(90)
```

Puoi trovare un elenco completo di tutte le funzioni della turtle supportate in Python 3.6 qui:
https://docs.python.org/3.6/library/turtle.html

Ora creiamo una nuova funzione per disegnare il lato di una figura.

```python
def side():  
    dan.forward(50)
    dan.right(90)

for i in range(4):
    side()
```

Nota che non abbiamo dovuto inserire l'import o l'inizializzazione dell'oggetto. Quello faceva parte dello stato del notebook.

Ora aggiungiamo un po' di colore e cambiamo lo spessore della penna da disegno a seconda che il numero del lato sia pari o dispari. Il percento è la funzione modulo.

```python
def side(edge):
    if edge % 2:
       dan.color('blue')
       dan.pensize(6)
    else:
       dan.color('red')
       dan.pensize(2)
    dan.forward(50)
    dan.right(90)

for i in range(4):
    side(i)
```

Ora aggiungiamo delle variabili e cambiamo il numero di lati. Qui creeremo un esagono a sei lati cambiando l'angolo a 60 e il range a 6.

```python
def side(edge, distance, angle):
    # this will be true for alternate edges
    if edge % 2:
       dan.color('blue')
       dan.pensize(6)
    else:
       dan.color('red')
       dan.pensize(2)
    dan.forward(distance)
    dan.right(angle)

# draw a six sided hexagon
dan.clear()
distance = 80
angle = 60
for i in range(6):
    side(i, distance, angle)
```
